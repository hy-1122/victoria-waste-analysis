## Cleaning and Standardization
### 1. Standardize Column Name

```python
# rename Population column
vlgas_df = vlgas_df.rename(columns={
    'Population': 'population',
    'kerbside_organics_premises_residential': 'premises_with_kerbside_organics_service_residential',
    'garbage_service_frequency': 'service_frequency_garbage',
    'kerbside_recycling_service_frequency': 'service_frequency_kerbside_recycling',
    'kerbside_organics_bin_frequency': 'service_frequency_kerbside_organics',
    'kerbside_glass_service_frequency': 'service_frequency_kerbside_glass',
    'kerbside_organics_fogo_included': 'has_fogo',
    'garbage_collected_total_tonnes_estimated': 'garbage_is_estimated'
})
```

### 2. Extract Start Year of Financial Year

```python
# extract start year of financial year
vlgas_df['start_year'] = vlgas_df['financial_year'].str[:4].astype(int)
vlgas_df['end_year'] = vlgas_df['financial_year'].str[-4:].astype(int)
```

```python
display(vlgas_df[vlgas_df['population'].isna()])
```

### 3. Convert Yes/No Columns to bool Columns 
#### FOGO Bin Column
`FOGO Services` column had numeric data leaked into it.

```python
# Check organics that are FOGO services
print('FOGO included in Organic Waste')
print('Null:')
print(vlgas_df['has_fogo'].isna().sum())
print(vlgas_df['has_fogo'].value_counts())
```

```python
vlgas_df['has_fogo'] = np.where(
    (vlgas_df['has_fogo'] == "No") |
    (vlgas_df['has_fogo'] == 0) |
    (vlgas_df['has_fogo'].isna()),
    False,
    True
)

vlgas_df['has_fogo'].value_counts()
```

#### Garbage Estimated Column

```python
vlgas_df['garbage_is_estimated'] = np.where(
    vlgas_df['garbage_is_estimated'] == "Yes",
    True,
    False
)
```

### 4. Split Organics waste collected from FOGO bin to new columns.

```python
# Clean FOGO service column
vlgas_df['kerbside_fogo_collected_tonnes'] = np.where(
    vlgas_df['has_fogo'] == True,
    vlgas_df['kerbside_organics_collected_tonnes'],
    np.nan
)

vlgas_df['kerbside_fogo_processed_tonnes'] = np.where(
    vlgas_df['has_fogo'] == True,
    vlgas_df['kerbside_organics_processed_tonnes'],
    np.nan
)

vlgas_df['kerbside_go_collected_tonnes'] = np.where(
    vlgas_df['has_fogo'] == False,
    vlgas_df['kerbside_organics_collected_tonnes'],
    np.nan
)

vlgas_df['kerbside_go_processed_tonnes'] = np.where(
    vlgas_df['has_fogo'] == False,
    vlgas_df['kerbside_organics_processed_tonnes'],
    np.nan
)
```

### 5. Service Frequency Standardization

```python
print(vlgas_df['service_frequency_garbage'].value_counts())
print(vlgas_df['service_frequency_kerbside_recycling'].value_counts())
print(vlgas_df['service_frequency_kerbside_organics'].value_counts())
print(vlgas_df['service_frequency_kerbside_glass'].value_counts())
```

```python
display(vlgas_df[vlgas_df['service_frequency_garbage'] == "weekly until the 08/04/2024 changed to fortnightly service"])

# Standardise the garbage service frequency for the affected rows
vlgas_df['service_frequency_garbage'] = vlgas_df['service_frequency_garbage'].replace({
    'weekly until the 08/04/2024 changed to fortnightly service': 'Weekly'
})
```

```python
# Standardise the glass service frequency for the affected rows
vlgas_df['service_frequency_kerbside_glass'] = vlgas_df['service_frequency_kerbside_glass'].replace({
    '4 weekly': 'Monthly',
    '4 weekly, 13 times per year': 'Monthly',
    '4 Weekly Once': 'Monthly',
    'every 4 weeks': 'Monthly',
    '8 weeks': 'Bi-monthly',
    'Every 8 weeks': 'Bi-monthly',
    'bimonthly': 'Bi-monthly',
})
```

```python
# Standardise the organics bin service frequency for the affected rows
vlgas_df['service_frequency_kerbside_organics'] = vlgas_df['service_frequency_kerbside_organics'].replace({
    'Biannual': 'Bi-annually',
    'Bi-annual': 'Bi-annually',
    'Annual': 'Annually',
    'twice monthly': 'Fortnightly',
    '4 weeks': 'Monthly',
    'First and Third Friday of each Month': 'Fortnightly',
    'On call': 'Other'
})
```

### 6. Missing Values

#### a. All Stream Waste Data Missing
Waste data from `Greater Geelong City Council` during `FY2021-2022` and `Hume City Council` during `FY2022-2023` is not recorded. These 2 rows are removed as it does not provide any information.

```python
missing_waste_data = vlgas_df[vlgas_df['garbage_collected_total_tonnes'].isna()]
display(missing_waste_data)
# remove rows with missing waste data
vlgas_df.drop(missing_waste_data.index, inplace=True)
```

#### b. Population data missing
`Population` data from `Boroondara City Council` and `Maribyrnong City Council` in `FY2020-2021` is not recorded.The 2 values is replaced with data from Australian Bureau of Statistics (ABS) Census 2021.

```python
display(vlgas_df[vlgas_df['population'].isna()])
```

```python
vlgas_df.loc[vlgas_df['population'].isna(), 'population'] = [167900, 85209]
```

### 7. Add Column: has_glass_service, has_recycling_service

```python
vlgas_df['has_glass_service'] = np.where(
    vlgas_df['kerbside_glass_total_collected_tonnes'].notna() & vlgas_df['kerbside_glass_total_collected_tonnes'] > 0,
    True,
    False
)

vlgas_df['has_organics_service'] = np.where(
    vlgas_df['kerbside_organics_collected_tonnes'].notna() & vlgas_df['kerbside_organics_collected_tonnes'] > 0,
    True,
    False
)
```

```python
vlgas_df['has_organics_service'] = (
    vlgas_df.groupby('council')['has_organics_service']
    .transform(lambda x: x.cummax())
)
```

## City of Melbourne Dataset

**Source:** `Dataset/Raw/waste-collected-per-month.xlsx`
Monthly waste collected by stream for the City of Melbourne council area, April 2009 – March 2020.
Used exclusively for the Part 3 city-level case study in the forecasting model.

**Cleaning steps:**
1. Parse `date` column from object to `datetime64`
2. Derive integer `year` and `month_num` columns
3. Assert expected columns are present
4. Integrity check: `hardwaste_total == hardwaste_to_landfill + hardwaste_recovered` (within floating-point tolerance)
5. Assert no nulls and no negative tonnage values
6. Export to `Dataset/Cleaned/cleaned_melbourne.csv`


```python
melb_df = pd.read_excel('Dataset/Raw/waste-collected-per-month.xlsx')
print("Shape:", melb_df.shape)
melb_df.info()
```

```python
melb_df.describe()
```

```python
# ── 1. Parse date ────────────────────────────────────────────────────────────
melb_df['date'] = pd.to_datetime(melb_df['date'])

# ── 2. Derive year and month_num for convenience ──────────────────────────────
melb_df['year']      = melb_df['date'].dt.year
melb_df['month_num'] = melb_df['date'].dt.month

# ── 3. Assert expected columns ────────────────────────────────────────────────
EXPECTED_COLS = {
    'date', 'year', 'month_num', 'month',
    'residential', 'public_litter_bins', 'dumped_rubbish', 'street_sweepings',
    'mattresses', 'commingled_recycling', 'cardboard',
    'hardwaste_to_landfill', 'hardwaste_recovered', 'hardwaste_total',
    'green_waste',
}
missing = EXPECTED_COLS - set(melb_df.columns)
assert not missing, f"Missing columns after cleaning: {missing}"
print("Column assertion passed.")

# ── 4. Integrity check: hardwaste_total == to_landfill + recovered ────────────
hw_computed = melb_df['hardwaste_to_landfill'] + melb_df['hardwaste_recovered']
hw_diff     = (melb_df['hardwaste_total'] - hw_computed).abs()
n_fail      = (hw_diff > 0.01).sum()   # allow 10g floating-point tolerance
if n_fail:
    print(f"WARNING: {n_fail} rows where hardwaste_total != to_landfill + recovered (diff > 0.01 t)")
    print(melb_df[hw_diff > 0.01][['date', 'hardwaste_to_landfill', 'hardwaste_recovered',
                                    'hardwaste_total']].to_string())
else:
    print(f"Hardwaste integrity check passed ({len(melb_df)} rows).")

# ── 5. Null and negative checks ───────────────────────────────────────────────
n_nulls = melb_df.isnull().sum().sum()
assert n_nulls == 0, f"Unexpected nulls: {melb_df.isnull().sum()[melb_df.isnull().sum() > 0]}"
print(f"Null check passed (0 nulls).")

tonnage_cols = [
    'residential', 'public_litter_bins', 'dumped_rubbish', 'street_sweepings',
    'mattresses', 'commingled_recycling', 'cardboard',
    'hardwaste_to_landfill', 'hardwaste_recovered', 'hardwaste_total', 'green_waste',
]
for col in tonnage_cols:
    n_neg = (melb_df[col] < 0).sum()
    assert n_neg == 0, f"Negative values in '{col}': {n_neg} rows"
print(f"Negative-value check passed ({len(tonnage_cols)} columns).")

# ── Summary ───────────────────────────────────────────────────────────────────
print(f"\nDate range : {melb_df['date'].min().date()} to {melb_df['date'].max().date()}")
print(f"Years      : {sorted(melb_df['year'].unique())}")
print(f"Rows       : {len(melb_df)}")
```

```python
# Reorder columns: identifiers first, then waste streams
col_order = [
    'date', 'year', 'month_num', 'month',
    'residential', 'public_litter_bins', 'dumped_rubbish', 'street_sweepings',
    'mattresses', 'commingled_recycling', 'cardboard',
    'hardwaste_to_landfill', 'hardwaste_recovered', 'hardwaste_total',
    'green_waste',
]
melb_df = melb_df[col_order]
melb_df.to_csv('Dataset/Cleaned/cleaned_melbourne.csv', index=False)
print(f"Exported cleaned_melbourne.csv  ({melb_df.shape[0]} rows x {melb_df.shape[1]} cols)")
melb_df.head()
```
