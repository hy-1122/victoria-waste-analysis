## Handle Missing Data

Fill early years missing data with 0.

```python
mask = vlgas_df['kerbside_recycling_total_collected_tonnes'].isna()

vlgas_df.loc[mask, 'premises_with_kerbside_recycling_service_residential'] = 0
vlgas_df.loc[mask, 'kerbside_recycling_total_recycled_tonnes'] = 0
vlgas_df.loc[mask, 'kerbside_recycling_total_collected_tonnes'] = 0
```

Interpolate other missing data.

```python
vlgas_df = vlgas_df.sort_values(by=['council', 'start_year']).reset_index(drop=True)

vlgas_df['premises_with_kerbside_recycling_service_residential'] = vlgas_df.groupby('council')['premises_with_kerbside_recycling_service_residential'].transform(
        lambda group: group.interpolate(method='linear', limit_direction='both')
    )

vlgas_df['kerbside_recycling_total_recycled_tonnes'] = vlgas_df.groupby('council')['kerbside_recycling_total_recycled_tonnes'].transform(
        lambda group: group.interpolate(method='linear', limit_direction='both')
    )
```

```python
# columns of all waste streams
all_waste = [
    'garbage_collected_total_tonnes',
    'kerbside_recycling_total_collected_tonnes',
    'kerbside_organics_collected_tonnes',
    'dropoff_garbage_collected_tonnes',
    'dropoff_recycling_collected_tonnes',
    'kerbside_glass_total_collected_tonnes',
    'hardwaste_collected_tonnes'
]

# columns of garbage waste
garbage_waste = [
    'garbage_collected_total_tonnes',
    'dropoff_garbage_collected_tonnes'
]

# columns of recycling waste
recycling_waste = [
    'kerbside_recycling_total_collected_tonnes',
    'dropoff_recycling_collected_tonnes',
]

# compute recovered hardwaste
vlgas_df['hardwaste_recovered'] = (vlgas_df['hardwaste_collected_tonnes'].fillna(0) - vlgas_df['hardwaste_disposed_tonnes'].fillna(0)).clip(lower=0)

# columns from kerbside collection
kerbside_total_waste_col = [
    'garbage_collected_total_tonnes',
    'kerbside_recycling_total_collected_tonnes',
    'kerbside_organics_collected_tonnes'
]

# columns of diverted waste from kerbside collection
diverted_waste_cols = [
    'kerbside_recycling_total_recycled_tonnes',
    'kerbside_organics_processed_tonnes',
]

garbage_services_cols = [
    'premises_with_garbage_service_residential',
    'premises_with_garbage_service_nonresidential'
]

recycling_services_cols = [
    'premises_with_kerbside_recycling_service_residential',
    'premises_with_kerbside_recycling_service_nonresidential'
]

# add new columns for different total of waste generation and diversion rates
vlgas_df['total_waste_generation'] = vlgas_df[all_waste].sum(axis=1)
vlgas_df['garbage_waste'] = vlgas_df[garbage_waste].sum(axis=1)
vlgas_df['recycling_waste'] = vlgas_df[recycling_waste].sum(axis=1)
vlgas_df['kerbside_diverted_waste'] = vlgas_df[diverted_waste_cols].sum(axis=1)
vlgas_df['kerbside_total_waste'] = vlgas_df[kerbside_total_waste_col].sum(axis=1)
vlgas_df['kerbside_diversion_rate'] = vlgas_df['kerbside_diverted_waste'] / vlgas_df['kerbside_total_waste']
vlgas_df['premises_with_garbage_services'] = vlgas_df[garbage_services_cols].fillna(0).sum(axis=1)
vlgas_df['premises_with_recycling_services'] = vlgas_df[recycling_services_cols].fillna(0).sum(axis=1)
```

## Aggregate Waste Data from All Councils

```python
# aggregate data by year and compute annual waste generation and diversion rates
annual_waste = (
        vlgas_df.groupby("start_year").agg(
            population     = ("population", "sum"),
            total_waste_generation    = ("total_waste_generation", "sum"),
            garbage_collected        = ("garbage_waste", "sum"),
            recycling_collected      = ("recycling_waste", "sum"),
            recycling_actual_recycled       = ("kerbside_recycling_total_recycled_tonnes", "sum"),
            organics_collected       = ("kerbside_organics_collected_tonnes", "sum"),
            fogo_collected          = ("kerbside_fogo_collected_tonnes", "sum"),
            go_collected            = ("kerbside_go_collected_tonnes", "sum"),
            organics_processed = ("kerbside_organics_processed_tonnes", "sum"),
            glass_collected          = ("kerbside_glass_total_collected_tonnes", "sum"),
            hardwaste_collected      = ("hardwaste_collected_tonnes", "sum"),
            hardwaste_recovered = ("hardwaste_recovered", "sum"),
            kerbside_total_waste = ("kerbside_total_waste", "sum"),
            kerbside_diverted_waste = ("kerbside_diverted_waste", "sum"),
            kerbside_diversion_rate   = ("kerbside_diversion_rate", "mean"),
            councils_with_fogo = ("has_fogo", "sum"),
            councils_with_glass = ("has_glass_service", "sum"),
            premises_with_garbage = ("premises_with_garbage_services", "sum"),
            premises_with_recycling = ("premises_with_recycling_services", "sum")
        ).fillna(0)
    )

waste_columns = [
    "total_waste_generation",
    "garbage_collected",
    "recycling_collected",
    "recycling_actual_recycled",
    "organics_collected",
    "fogo_collected",
    "go_collected",
    "organics_processed",
    "glass_collected",
    "hardwaste_collected",
    "hardwaste_recovered",
    "kerbside_total_waste",
    "kerbside_diverted_waste"
]

annual_waste[waste_columns] = (annual_waste[waste_columns] / 1000).round(2)
annual_waste["kerbside_diversion_rate"] = annual_waste['kerbside_diversion_rate'].round(4)
annual_waste["fogo_adoption_rate"] = (annual_waste["councils_with_fogo"] / 79).round(4)
annual_waste["fogo_usage_share"] = (annual_waste["fogo_collected"] /(annual_waste["fogo_collected"] +annual_waste["garbage_collected"])).fillna(0)

annual_waste['glass_collected'] = annual_waste['glass_collected'].replace(0, np.nan)

annual_waste['hardwaste_collected'] = annual_waste['hardwaste_collected'].replace(0, np.nan)
annual_waste["hardwaste_collected"] = annual_waste["hardwaste_collected"].transform(
            lambda group: group.interpolate(method='linear', limit_direction='both')
    )
```

```python
vwp_df = pd.read_csv('Dataset/Cleaned/cleaned_vwp.csv')
```

```python
vwp_df
```

```python
annual_vwp = (
        vwp_df.groupby("start_year").agg(
            disposal_full = ("disposal", "sum"),
            total_generation_full = ("total_generation", "sum")
        ).fillna(0)
    )

organics_full = vwp_df[vwp_df['material_type'] == "Organics"]
annual_organics_full = (
    organics_full.groupby("start_year").agg(
        disposal_organics = ("disposal", "sum"),
        total_generation_organics = ("total_generation", "sum")
    ).fillna(0)
)

glass_full = vwp_df[(vwp_df['material_type'] == "Glass") & (vwp_df['source_sector'] == "MSW")]
annual_glass_full = (
    glass_full.groupby("start_year").agg(
        disposal_glass = ("disposal", "sum"),
        total_generation_glass = ("total_generation", "sum")
    ).fillna(0)
)

annual_vwp['disposal_full'] = (annual_vwp['disposal_full'] / 1000).round(2)
annual_vwp['total_generation_full'] = (annual_vwp['total_generation_full'] / 1000).round(2)
annual_vwp['disposal_organics'] = (annual_organics_full['disposal_organics']/ 1000).round(2)
annual_vwp['total_generation_organics'] = (annual_organics_full['total_generation_organics'] / 1000).round(2)
annual_vwp['disposal_glass'] = (annual_glass_full['disposal_glass']/ 1000).round(2)
annual_vwp['total_generation_glass'] = (annual_glass_full['total_generation_glass'] / 1000).round(2)
annual_vwp['recovered_full'] = annual_vwp['total_generation_full'] - annual_vwp['disposal_full']
annual_vwp["diversion_rate"] = annual_vwp['recovered_full'] / annual_vwp['total_generation_full']
```

```python
annual_waste = pd.concat([annual_waste, annual_vwp], axis=1)
```

```python
annual_waste.to_csv('Dataset/ML/annual_waste.csv', index=True)
```
